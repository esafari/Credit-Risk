```python
data['Duration in months'].unique()
data['Housing'].unique()
data['Job'].unique()
print(data['Other debtors / guarantors'].unique())
```

```python
numeric_feature=[fe for fe in data.columns if data[fe].dtype!='O']
print(len(numeric_feature),numeric_feature)
Categorical_feature=[fe for fe in data.columns if data[fe].dtype=='O']
print(len(Categorical_feature),Categorical_feature)
```

```python
cr=data[numeric_feature].corr(method='spearman')
mask = np.array(cr)
mask[np.tril_indices_from(mask)] = False
fig, ax = plt.subplots(figsize = (15,12))
fig.set_size_inches(20,20)
sns.heatmap(cr, mask = mask, vmax = 0.9, square = True, annot = True)
```

```python
cr=data[numeric_feature].corr(method='pearson')
mask = np.array(cr)
mask[np.tril_indices_from(mask)] = False
fig, ax = plt.subplots(figsize = (15,12))
fig.set_size_inches(20,20)
sns.heatmap(cr, mask = mask, vmax = 0.9, square = True, annot = True)
```
